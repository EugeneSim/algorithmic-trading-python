## Backtest, Walk-Forward, Monte Carlo

```python
from src.backtest.engine import backtest_returns, backtest_summary
from src.walk_forward.testing import walk_forward_test
from src.monte_carlo.bootstrap import bootstrap_sharpe, monte_carlo_simulate
import pandas as pd
import numpy as np

# Sample returns (use yfinance in production)
np.random.seed(42)
dates = pd.date_range('2018-01-01', periods=252*3, freq='B')
rets = pd.DataFrame(np.random.randn(len(dates), 20).astype(np.float64) * 0.01, index=dates)

equity = backtest_returns(rets, rebalance_freq='M', initial_capital=100_000)
summary = backtest_summary(equity)
print('Backtest:', summary)

wf = walk_forward_test(rets.iloc[:, 0], train_months=24, test_months=12, step_months=12)
print('Walk-forward:', wf.head(2))

monthly = rets.resample('M').apply(lambda x: (1+x).prod()-1).iloc[:, 0]
mean_s, lo, hi = bootstrap_sharpe(monthly, n_simulations=200)
print('Bootstrap Sharpe: mean=%.3f, 5%%=%.3f, 95%%=%.3f' % (mean_s, lo, hi))
```
